# Evaluación del Recomendador Híbrido

Métricas: Precisión@K, Diversidad, Cobertura del catálogo.

```python
import sys
sys.path.insert(0, '../src')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from data_loader import load_clean, AUDIO_FEATURES
from content_model import build_feature_matrix, get_content_recommendations
from collab_model import load_model, get_collab_recommendations
from hybrid import get_hybrid_recommendations
from synthetic_users import generate_interactions

df = load_clean()
feature_matrix = build_feature_matrix(df)
interactions = generate_interactions(df)
model = load_model()
print(f'Dataset: {len(df)} canciones | Usuarios: {interactions["user_id"].nunique()}')
```

```python
# --- Precision@K ---
# Para cada usuario de prueba, comparamos las recomendaciones con su arquetipo
def precision_at_k(user_id, archetype, df, feature_matrix, interactions, model, alpha, k=10):
    user_tracks = interactions[interactions['user_id'] == user_id]['track_id'].values
    if len(user_tracks) == 0:
        return None
    seed_track = user_tracks[0]
    try:
        recs = get_hybrid_recommendations(
            track_id=seed_track, user_id=user_id,
            df=df, feature_matrix=feature_matrix,
            interactions=interactions, model=model,
            alpha=alpha, n=k
        )
    except Exception:
        return None
    # Precision: qué % de recomendaciones coincide con el perfil del usuario
    user_genres = df[df['track_id'].isin(user_tracks)]['track_genre'].value_counts()
    top_genres = set(user_genres.head(3).index)
    rec_genres = recs['track_genre'].values
    hits = sum(1 for g in rec_genres if g in top_genres)
    return hits / k

test_users = interactions.groupby('archetype').apply(lambda x: x['user_id'].unique()[:5]).explode().values
results = []
for alpha in [0.0, 0.5, 1.0]:
    for uid in test_users:
        arch = interactions[interactions['user_id']==uid]['archetype'].iloc[0]
        p = precision_at_k(uid, arch, df, feature_matrix, interactions, model, alpha=alpha, k=10)
        if p is not None:
            results.append({'user_id': uid, 'archetype': arch, 'alpha': alpha, 'precision@10': p})

results_df = pd.DataFrame(results)
print(results_df.groupby(['alpha', 'archetype'])['precision@10'].mean().round(3))
```

```python
# --- Visualización Precision@K por alpha ---
fig, ax = plt.subplots(figsize=(10, 5))
pivot = results_df.groupby(['alpha', 'archetype'])['precision@10'].mean().unstack()
pivot.plot(kind='bar', ax=ax, colormap='viridis')
ax.set_title('Precision@10 por arquetipo y peso híbrido (alpha)')
ax.set_xlabel('Alpha (0=colaborativo, 1=contenido)')
ax.set_ylabel('Precision@10')
ax.legend(title='Arquetipo', bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.savefig('../data/processed/eval_precision_at_k.png', dpi=100)
plt.show()
```

```python
# --- Cobertura del catálogo ---
all_recommended = set()
sample_users = interactions['user_id'].unique()[:50]
sample_tracks = df['track_id'].sample(50, random_state=42).values

for uid, tid in zip(sample_users, sample_tracks):
    try:
        recs = get_hybrid_recommendations(
            track_id=tid, user_id=uid,
            df=df, feature_matrix=feature_matrix,
            interactions=interactions, model=model,
            alpha=0.5, n=10
        )
        all_recommended.update(recs['track_id'].values)
    except Exception:
        pass

coverage = len(all_recommended) / len(df) * 100
print(f'Cobertura del catálogo (50 consultas): {coverage:.1f}% ({len(all_recommended)} canciones únicas recomendadas de {len(df)})')
```

```python
# --- Diversidad intra-lista ---
from sklearn.metrics.pairwise import cosine_similarity as cos_sim
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
norm_matrix = scaler.fit_transform(df[AUDIO_FEATURES].values)

diversities = []
for tid in df['track_id'].sample(30, random_state=1).values:
    try:
        recs = get_content_recommendations(tid, df, feature_matrix, n=10)
        idxs = [df.index[df['track_id']==t][0] for t in recs['track_id'].values if t in df['track_id'].values]
        if len(idxs) < 2:
            continue
        vecs = norm_matrix[idxs]
        sim_matrix = cos_sim(vecs)
        n = len(idxs)
        avg_sim = (sim_matrix.sum() - n) / (n * (n - 1))
        diversities.append(1 - avg_sim)
    except Exception:
        pass

print(f'Diversidad intra-lista promedio (modelo contenido): {np.mean(diversities):.3f}')
print('(1.0 = máx. diversidad, 0.0 = mín. diversidad)')
```
